```python
# ── Cell 6: Verify Exported TFLite File ─────────────────────────────────
import os, glob

weights_dir = '/content/drive/MyDrive/Fish-Freshness-1/training_results/aqua_scan_v1/weights'
tflite_files = glob.glob(os.path.join(weights_dir, '**/*.tflite'), recursive=True)

if tflite_files:
    for f in tflite_files:
        size_mb = os.path.getsize(f) / 1e6
        print(f"✅  {f}")
        print(f"    Size: {size_mb:.2f} MB")
else:
    print("❌ No .tflite files found — check export step above.")
```

```python
import yaml

# Path to the args.yaml file
args_path = '/content/drive/MyDrive/Fish-Freshness-1/Fish-Freshness-2/training_results/aqua_scan_v1/args.yaml'
new_data_path = '/content/drive/MyDrive/Fish-Freshness-1/Fish-Freshness-2/data.yaml'

# Load, modify, and save the yaml
with open(args_path, 'r') as f:
    args_data = yaml.safe_load(f)

args_data['data'] = new_data_path

with open(args_path, 'w') as f:
    yaml.dump(args_data, f)

print(f"Successfully updated 'data' path in {args_path} to: {new_data_path}")
```
